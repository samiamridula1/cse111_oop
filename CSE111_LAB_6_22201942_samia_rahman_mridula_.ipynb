```python
class SultansDine:
    total_branches = 0
    total_sell = 0
    branch_list = []


    def __init__(self, name):
        self.name = name
        self.sell_quantity = 0
        self.branch_sell = 0
        SultansDine.total_branches += 1


    def sellQuantity(self, quantity):
        self.sell_quantity = quantity
        if quantity < 10:
            self.branch_sell = quantity * 300
        elif quantity < 20:
            self.branch_sell = quantity * 350
        else:
            self.branch_sell = quantity * 400
        self.individualList = []
        self.individualList.append(self.name)
        self.individualList.append(self.branch_sell)
        SultansDine.branch_list.append(self.individualList)

        SultansDine.total_sell += self.branch_sell


    def branchInformation(self):
        print(f'Branch Name: {self.name}')
        print(f'Branch Sell: {self.branch_sell} Taka')

    @classmethod
    def details(cls):
        print(f'Total Number of branch(s): {cls.total_branches}')
        print(f'Total Sell: {cls.total_sell} Taka')
        if cls.total_branches != 0:
            for i in cls.branch_list:
                bname, bsell = i
                print(f'Branch Name: {bname}, Branch Sell:{bsell}')
                percentage = (bsell / SultansDine.total_sell) * 100
                print(f"Branch consists of total sell's: {format(percentage)}.2f%")


SultansDine.details()
print('########################')
dhanmondi = SultansDine('Dhanmondi')
dhanmondi.sellQuantity(25)
dhanmondi.branchInformation()
print('-----------------------------------------')
SultansDine.details()
print('========================')
baily_road = SultansDine('Baily Road')
baily_road.sellQuantity(15)
baily_road.branchInformation()
print('-----------------------------------------')
SultansDine.details()
print('========================')
gulshan = SultansDine('Gulshan')
gulshan.sellQuantity(9)
gulshan.branchInformation()
print('-----------------------------------------')
SultansDine.details()
```
